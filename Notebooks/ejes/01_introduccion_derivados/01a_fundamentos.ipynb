### Money Market (1/4)

El mercado monetario (Money Market) es el conjunto de mercados financieros donde se intercambian activos con plazo de amortización corto (menor a 18 meses), bajo riesgo y elevada liquidez.

Ejemplo: bono (muy) corto plazo soberano. Pedir dinero prestado sin costo/riesgo. Suelen denominarse libres de riesgo (risk free).

### Arbitraje — There is no free lunch

En economía y finanzas, **arbitraje** es la práctica de tomar ventaja de una diferencia de precio entre dos o más mercados: realizar una combinación de transacciones complementarias que capitalizan el desequilibrio de precios.

**Hipótesis fundamental:** No existen posibilidades de arbitraje.

Las oportunidades existen en la realidad, pero las ventanas son (deberían ser) pequeñas.

Existen oportunidades de arbitraje si existe un portafolio que cumpla simultáneamente:

- Valor inicial = 0 (empiezo gratis)
- Valor final ≥ 0 con probabilidad 1, y > 0 con probabilidad positiva (termino positivo)

Es decir: **empiezo gratis, termino positivo**.

## 3) Números importantes

**En todo momento:**

| Concepto | Definición |
|----------|------------|
| **Precio** | Número que se observa en un mercado (organizado o OTC); lo que se pagaría/recibiría para negociar |
| **Valor Mark-to-Market** | Estimación de lo que se podría pagar/recibir — requiere **modelos** |
| **Prima (Premium)** | Lo que realmente se paga/recibe al negociar |

**A vencimiento:**

| Concepto | Definición |
|----------|------------|
| **Payoff** | Valor a tiempo final del contrato — **determinístico** |
| **Ganancia / PnL** | Ganancia o pérdida total, incluyendo prima y valor del derivado (o payoff realizado) |

## 4) Derivados — qué son

Un **derivado** es un contrato cuyo valor depende del precio de un **activo subyacente** (acción, índice, tasa, commodity, etc.).

**Clasificación:**

- **Lineales**: Forward, Futuros, Swaps — el valor evoluciona de forma lineal con el subyacente
- **No lineales**: Opciones — el valor evoluciona de forma no lineal
- **Otros**: Productos estructurados, híbridos (ej. convertibles)
