## 1) Derivados lineales — Forward

Un **forward** es un contrato para comprar/vender un activo en una fecha futura a un precio pactado (precio forward $F$).

**Payoff** (posición larga): al vencimiento $T$, el payoff es $S_T - F$.

**Valuación:** En ausencia de arbitraje, el precio forward que elimina arbitraje es $F = S_0 \, e^{(r-q)T}$ (con tasa $r$ y dividend yield $q$).

Un forward sintético se puede replicar con: **Call(K) − Put(K)** = $S - K$.

```python
import sys
sys.path.append('../../..')

import numpy as np
import matplotlib.pyplot as plt

from Codigo.analytics.payoffs import payoff_forwardSintetico, payoff_call, payoff_put
```

```python
# Payoff Forward sintético: Call(K) - Put(K) = S - K
K = 100
S = np.linspace(50, 150, 200)

payoff_fwd = payoff_forwardSintetico(S, K)

plt.figure(figsize=(7, 4))
plt.plot(S, payoff_fwd, 'b-', linewidth=2, label='Forward (long)')
plt.axvline(K, color='gray', linestyle='--', alpha=0.7, label=f'K = {K}')
plt.axhline(0, color='gray', linestyle='-', alpha=0.7)
plt.xlabel('Precio subyacente S al vencimiento')
plt.ylabel('Payoff')
plt.title('Payoff contrato forward (posición larga)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5) Opciones — definición y parangones

**Definición:** Una opción es un contrato que da al dueño el **derecho, pero no la obligación**, de negociar un activo (subyacente) por un precio (strike) en una fecha futura (expiración).

- **Call**: derecho a **comprar**
- **Put**: derecho a **vender**

**Jerga:** Underlying, Strike Price, Expiry/Maturity.

**Parangón Call:** Depósito para compra de vivienda. Pagamos depósito; tenemos derecho (no obligación) de comprar. Si no compramos, perdemos el depósito.

**Parangón Put:** Seguro de vivienda. Pagamos prima; recibimos protección. Si hay daños, la aseguradora paga reparaciones.

### ¿Para qué se usan?

- **Comprador**: riesgo limitado (derecho, no obligación). **Vendedor**: riesgo mayor.
- Cobertura contra movimientos de precios
- Beneficio ante movimientos del subyacente
- Apalancamiento y replicación con posiciones sintéticas
- Estrategias combinando varias opciones
