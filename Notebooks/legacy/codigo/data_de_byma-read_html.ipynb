# Bonus: read_html

```python
#url = 'http://www.rava.com/precios/panel.php?m=OPC'
#url = 'http://clasico.rava.com/precios/panel.php?m=OPC'
#url = 'http://datos.rava.com/cotizaciones/opciones'
url = 'https://iol.invertironline.com/mercado/cotizaciones/argentina/opciones/todas'
#url = 'https://www.rava.com/cotizaciones/opciones'


from bs4 import BeautifulSoup
import requests
import pandas as pd
page = requests.get(url)
soup = BeautifulSoup(page.text, 'lxml')
opciones_html = soup.find('table', {'id': 'cotizaciones'})
```

```python
import pandas as pd
dfs = pd.read_html(url)

# dfs es una lista con un dataframe por cada tabla encontrada en la url
# si hay varias tablas en el html devolverá varios dataframes

# de la doc:
# https://pandas.pydata.org/docs/reference/api/pandas.read_html.html
# [...] This function searches for <table> elements and only for <tr> and <th> rows and 
# <td> elements within each <tr> or <th> element in the table. 
# <td> stands for “table data” [...]

print(len(dfs))
df = dfs[0]
df
```

### Varias maneras de especificar el acceso

```python
# con el html ya cargado, como hicimos con opciones_html
pd.read_html(str(opciones_html))[0]
```

```python
# por un valor especifico que conocemos dentro de la tabla
pd.read_html(url, match="YPFV920.OC", flavor="lxml")[0]
```

```python
# por id de la tabla
pd.read_html(url, attrs={"id": "cotizaciones"})[0]
```
