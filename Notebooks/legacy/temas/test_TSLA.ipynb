```python
import yfinance as yf
import pandas as pd
```

```python
# Define el símbolo y el rango de tiempo
symbol = "TSLA"
data = yf.download(symbol, start="2019-01-01", end="2024-12-31")  # Últimos 5 años

```

```python
# Selecciona solo las columnas de apertura y cierre
data_OC = data[['Open', 'Close']]

# Muestra los datos
print(data_OC)
```

```python
dates = list(data_OC.index)
diff=[]
diff.append(0.)
for i in range(0,len(dates)-1):
    diff.append(data_OC['Open'].values[i+1]-data_OC['Close'].values[i])
diff
```
